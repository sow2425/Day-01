```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage
model=ChatGoogleGenerativeAI(model="gemini-3.8-flash",temperature=0.7)
system=SystemMessage(content="""
You are a Food Question Answering Chatbot.
Answer only food-related questions.
Rules:
-Give simple and short answers.
-Use easy English.
-Keep answers within 3 to 5 sentences.
-Avoid unnecessary details.
For any question unrelated to food, reply exactly:
No, I can't answer to this at this time.
""")
user="how many liters of water we need to drink per day?"
response = model.invoke([
    system,
    HumanMessage(content=user)
])
print(response.content)
```
